```python
import pandas as pd

path = "../data/raw/Mobile_Parts_Wholesale_Dataset.xlsx"
sales = pd.read_excel(path, sheet_name="Sales_Transactions")
sales['order_date'] = pd.to_datetime(sales['order_date'])

print(sales.shape)
sales.head()
```

```python
daily_demand = sales.groupby('order_date').agg(
    total_qty=('quantity', 'sum'),
    total_revenue=('total_amount_inr', 'sum')
).reset_index()

print(daily_demand.shape)
daily_demand.head()
```

```python
full_range = pd.date_range(start=daily_demand['order_date'].min(),
                            end=daily_demand['order_date'].max(), freq='D')
daily_demand = daily_demand.set_index('order_date').reindex(full_range, fill_value=0).reset_index()
daily_demand = daily_demand.rename(columns={'index': 'order_date'})

print(daily_demand.shape)
daily_demand.head()
```

```python
daily_demand.to_csv("../data/processed/daily_demand.csv", index=False)
print("Saved daily_demand.csv")
```
